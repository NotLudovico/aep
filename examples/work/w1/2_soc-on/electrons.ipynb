```python
import warnings; warnings.filterwarnings("ignore", message="Using `tqdm.autonotebook.tqdm`")
from abipy.abilab import abiopen
```

```python
path = "./outdata/electrons_DS2_GSR.nc"
with abiopen(path) as file:
    bands = file.ebands
```

```python
fig = bands.plot(title="W BCC - SOC on", ylims=(-5, 10))
fig.savefig("./figures/ebnd.pdf")
```

```python
path = "./outdata/electrons_DS3_GSR.nc"
with abiopen(path) as file:
    dos = file.ebands.get_edos(method="gaussian", step=0.01, width=0.1)
```

```python
fig = dos.plot(e0=19.380276149736126, title="W BCC - SOC on", xlims=(-5, 10), ylims=(0, 4))
fig.savefig("./figures/edos.pdf")
```
